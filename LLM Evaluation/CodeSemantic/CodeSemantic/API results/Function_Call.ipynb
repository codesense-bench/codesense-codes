```python
import json
import inspect
import re
import pydoc
from importlib import import_module

# Pre-import common modules to allow eval() to work
COMMON_MODULES = ["os", "re", "json", "sys", "pathlib", "numpy", "pandas", "torch"]
for module in COMMON_MODULES:
    try:
        globals()[module] = import_module(module)
    except ImportError:
        pass

def extract_api_name(statement):
    """Extract the API name from a statement, ignoring variable assignments."""
    match = re.search(r"([a-zA-Z0-9_.]+)\(", statement)
    return match.group(1) if match else None

def get_api_description(api_name):
    """Fetch description of an API if available."""
    try:
        # Case 1: Qualified name (module.function)
        if '.' in api_name:
            parts = api_name.split('.')
            module_name = '.'.join(parts[:-1])
            func_name = parts[-1]
            module = import_module(module_name)
            func = getattr(module, func_name)
        # Case 2: Standalone function
        else:
            func = eval(api_name)
            module = inspect.getmodule(func)
        
        doc = pydoc.getdoc(func)
        if doc:
            module_name = module.__name__ if module else "builtins"
            return f"Module: {module_name}\nDescription: {doc}"
    
    except (ImportError, AttributeError, NameError):
        pass  # Not resolvable
    
    return "No public documentation found."

# Path to your JSONL file
jsonl_file = '/home/XXX/CodeSemantic/CodeSemantic/dataset/statement_prediction_dataset.jsonl'

with open(jsonl_file, 'r') as f:
    for line in f:
        data = json.loads(line)
        if data.get("Statement Type") == "API":
            statement = data.get("Selected Statement")
            print("\nStatement:")
            print(statement)
            
            api_name = extract_api_name(statement)
            if api_name:
                print(f"\nAPI Name: {api_name}")
                try:
                    description = get_api_description(api_name)
                    print(f"Documentation:\n{description}")
                except Exception as e:
                    print(f"Error fetching docs: {str(e)}")
            else:
                print("Could not extract API name.")
            print("-" * 50)
```

```python
import ast
import inspect
from typing import List, Dict, Optional

def get_callees(func_obj) -> List[str]:
    """Extract names of functions/methods called by the given function."""
    try:
        source = inspect.getsource(func_obj)
        tree = ast.parse(source)
        callees = set()
        
        # Walk through the AST to find function calls
        for node in ast.walk(tree):
            if isinstance(node, ast.Call):
                if isinstance(node.func, ast.Name):
                    callees.add(node.func.id)  # e.g., foo()
                elif isinstance(node.func, ast.Attribute):
                    callees.add(node.func.attr)  # e.g., obj.method()
        
        return sorted(callees)
    except (TypeError, OSError):
        return []  # Built-in/C-compiled functions have no source

def get_callee_documentation(callee_name: str, func_obj) -> Optional[Dict[str, str]]:
    """Get docs for a callee if it's in the same module."""
    try:
        module = inspect.getmodule(func_obj)
        if module and hasattr(module, callee_name):
            callee_obj = getattr(module, callee_name)
            doc = pydoc.getdoc(callee_obj)
            return {
                "name": callee_name,
                "module": module.__name__,
                "description": doc or "No docs available."
            }
    except Exception:
        pass
    return None

def get_full_api_docs(api_name: str) -> Dict[str, object]:
    """Enhanced to include callee documentation."""
    try:
        # Get the base API docs (original functionality)
        docs = {"api": api_name, "documentation": get_api_description(api_name)}
        
        # Resolve the function object
        if '.' in api_name:
            module_name, func_name = api_name.rsplit('.', 1)
            module = import_module(module_name)
            func_obj = getattr(module, func_name)
        else:
            func_obj = eval(api_name)
        
        # Get callees and their docs
        callees = get_callees(func_obj)
        docs["callees"] = []
        
        for callee in callees:
            callee_doc = get_callee_documentation(callee, func_obj)
            if callee_doc:
                docs["callees"].append(callee_doc)
        
        return docs
    
    except Exception as e:
        return {"error": str(e)}


# Path to your JSONL file
jsonl_file = '/home/XXX/CodeSemantic/CodeSemantic/dataset/statement_prediction_dataset.jsonl'

with open(jsonl_file, 'r') as f:
    for line in f:
        data = json.loads(line)
        if data.get("Statement Type") == "API":
            statement = data.get("Selected Statement")
            print("\nStatement:")
            print(statement)
            
            api_name = extract_api_name(statement)
            if api_name:
                print(f"\nAPI Name: {api_name}")
                try:
                    description = get_api_description(api_name)
                    print(f"Documentation:\n{description}")
                except Exception as e:
                    print(f"Error fetching docs: {str(e)}")
            else:
                print("Could not extract API name.")
            print("-" * 50)
```
