```python
import json

input_file = '/home/XXX/CodeSemantic/CodeSemantic/dataset/statement_prediction_dataset.jsonl'
output_file = 'boolean_prediction_dataset.jsonl'

with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
    for line in infile:
        try:
            data = json.loads(line)
            
            # Check if statement type is "Branch"
            if data.get("Statement Type") == "Branch":
                selected_stmt = data.get("Selected Statement", "").lstrip()
                
                # Check if selected statement starts with "if"
                if selected_stmt.startswith("if"):
                    # Strip "if" and any following whitespace/parentheses
                    condition = selected_stmt[2:].lstrip()  # Remove "if"
                    
                    # Remove parentheses if they wrap the entire condition
                    if condition.startswith('(') and condition.endswith(')'):
                        condition = condition[1:-1].strip()
                    
                    # Update the data and write to output
                    data["Selected Statement"] = condition
                    outfile.write(json.dumps(data) + '\n')
                    
        except json.JSONDecodeError:
            print("Error decoding JSON, skipping line")
            continue
```

```python
import json

input_file = '/home/XXX/CodeSemantic/CodeSemantic/dataset/statement_prediction_dataset.jsonl'
output_file = 'api_prediction_dataset.jsonl'

with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
    for line in infile:
        try:
            data = json.loads(line)
            
            # Check if statement type is "API"
            if data.get("Statement Type") == "API":
                # Write the data directly to output (no need to modify the statement)
                outfile.write(json.dumps(data) + '\n')
                
        except json.JSONDecodeError:
            print("Error decoding JSON, skipping line")
            continue
```
