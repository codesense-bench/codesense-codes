```python
from sklearn.model_selection import train_test_split
import json

with open('/home/XXX/CodeSemantic/CodeSemantic/dataset/loop_iteration_dataset_python.jsonl', 'r') as f:
    data = [json.loads(line) for line in f]

first_split_1, first_split_2 = train_test_split(data, test_size=0.5, random_state=42)


second_split_1, second_split_2 = train_test_split(data, test_size=0.5, random_state=123)
```

```python
def save_to_jsonl(data, filename):
    with open(filename, 'w') as f:
        for item in data:
            f.write(json.dumps(item) + '\n')

save_to_jsonl(first_split_1, 'first_split_1.jsonl')
save_to_jsonl(first_split_2, 'first_split_2.jsonl')
save_to_jsonl(second_split_1, 'second_split_1.jsonl')
save_to_jsonl(second_split_2, 'second_split_2.jsonl')
```
